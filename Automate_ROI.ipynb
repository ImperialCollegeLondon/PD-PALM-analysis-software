### Extracting ROI data from the experiment data script
This Jupyter Notebook script will extract the ROI data from the experiment data and save it as a separate CSV file. 

User guide:
1. Select desired side lenght of the ROI
2. Write in the path to folder containing the experiment and ROI data
3. Select if the script should save the files within a folder inside the output folder and if the experiment name should be used as a prefix for the file names
4. Run the notebook

```python
import os
import pandas as pd 
from glob import glob
from automation_utils import * # Importing all functions from automation_utils file


length = 5000

input_folder = '/Users/Katie/Library/CloudStorage/OneDrive-TheFrancisCrickInstitute/EGFR Alix collab project/2026-06-18_palm_repeat4/DNM2 no egf'
output_folder =  '/Users/Katie/Library/CloudStorage/OneDrive-TheFrancisCrickInstitute/EGFR Alix collab project/2026-06-18_palm_repeat4/DNM2 no egf'
save_in_nested_folder = False # Will create a folder with experiment name within the output folder, If false, files will be saved directly in the output folder
save_files_with_experiment_name = True # Will save files with experiment name, If false, files will be saved with the original file name



# -------- Rare settings -------- #
overwrite = False # If true, will overwrite files with the same name (will happen if the script is run multiple times with with same data and settings)
units_to_extract = 'nm'


qucik_check(save_in_nested_folder,save_files_with_experiment_name)
```

```python

# -------- Running the pipeline -------- #
output_folder_use  = output_folder
if input_folder[-1] != '/':
    input_folder += '/'
    
files = sorted(glob(os.path.join(input_folder, '*.csv')))  #It's better to add a sorted function here to sort the returned filenames
ROI_files = [f for f in files if 'ROI' in f]
data_files = [f for f in files if 'ROI' not in f]
for data_file, ROI_file in zip(data_files, ROI_files):
    print(data_file,'<--->', ROI_file)
    pass
# Above code can be updated to fit another file structure, for example, if the data and ROI files are in separate folders, or if multiple data files and ROI are in the same folder, etc.


for data_file, ROI_file in zip(data_files, ROI_files):
    run_squares_pipeline(
        data_file = data_file, 
        ROI_file = ROI_file, 
        output_folder = output_folder_use, 
        save_in_nested_folder= save_in_nested_folder, 
        length= length, 
        unit_to_extract= units_to_extract,
        save_files_with_experiment_name= save_files_with_experiment_name, 
        overwrite= overwrite)
```
